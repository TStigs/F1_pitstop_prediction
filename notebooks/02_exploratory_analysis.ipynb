# 02. Exploratory Data Analysis (EDA)
Analyze feature distributions, pit stop probabilities across race progress, tire age, and compounds.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from src.data import load_raw_data
train, _, _ = load_raw_data('../data/raw')
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=train, x='RaceProgress', hue='PitNextLap', bins=30, kde=True, common_norm=False)
plt.title('Pit Stop Probability Across Race Progress')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=train, x='Compound', y='TyreLife', hue='PitNextLap')
plt.title('TyreLife vs Compound by PitNextLap Target')
plt.show()
```
